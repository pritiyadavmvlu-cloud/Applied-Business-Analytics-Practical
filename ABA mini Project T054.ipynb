```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Retail and wherehouse Sale.csv")

data = df[["RETAIL TRANSFERS", "RETAIL SALES"]].dropna()

x = data["RETAIL TRANSFERS"].values
y = data["RETAIL SALES"].values

# Regression coefficient and intercept
coefficient = np.cov(x, y, bias=True)[0, 1] / np.var(x)
intercept = y.mean() - coefficient * x.mean()

# Predicted values
y_pred = intercept + coefficient * x

# R2
r2 = 1 - np.sum((y - y_pred) ** 2) / np.sum((y - y.mean()) ** 2)

# MSE and RMSE
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)

print("Regression Equation:")
print("Retail Sales =", intercept, "+", coefficient, "* Retail Transfers")

print("\nRegression Coefficient:", coefficient)
print("R² Score:", r2)
print("MSE:", mse)
print("RMSE:", rmse)

# Prediction
transfer_value = 100
prediction = intercept + coefficient * transfer_value

print("\nPredicted Retail Sales for 100 Retail Transfers:", prediction)

# Scatter plot and regression line
plt.figure(figsize=(8,5))
plt.scatter(x, y)
plt.plot(x, y_pred)
plt.title("Retail Transfers vs Retail Sales")
plt.xlabel("Retail Transfers")
plt.ylabel("Retail Sales")
plt.show()
```

```python
# Decision Making Under Uncertainty

alternatives = {
    "Large Inventory": [(0.6, 80000), (0.4, 20000)],
    "Small Inventory": [(0.6, 50000), (0.4, 35000)]
}

for alternative, outcomes in alternatives.items():
    expected_value = 0
    
    for probability, payoff in outcomes:
        expected_value += probability * payoff
    
    print(alternative)
    print("Expected Value:", expected_value)
    print()

# Compare alternatives
ev_large = 0.6 * 80000 + 0.4 * 20000
ev_small = 0.6 * 50000 + 0.4 * 35000

print("Comparison:")
print("Large Inventory EV:", ev_large)
print("Small Inventory EV:", ev_small)

if ev_large > ev_small:
    print("\nHigher Expected Value: Large Inventory")
else:
    print("\nHigher Expected Value: Small Inventory")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Retail and wherehouse Sale.csv")

# Population
population = df["RETAIL SALES"].dropna().values

print("Population Size:", len(population))

np.random.seed(10)

sample_sizes = [30, 50, 100]

for size in sample_sizes:

    sample_means = []

    # Take 100 random samples
    for i in range(100):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means.append(sample.mean())

    sample_means = np.array(sample_means)

    mean_of_means = sample_means.mean()
    sd = sample_means.std(ddof=1)
    standard_error = population.std(ddof=1) / np.sqrt(size)

    print("\nSample Size:", size)
    print("Mean of Sample Means:", mean_of_means)
    print("Standard Deviation:", sd)
    print("Standard Error:", standard_error)

    # Histogram
    plt.figure(figsize=(8,5))
    plt.hist(sample_means, bins=15)
    plt.title("Sampling Distribution of Sample Means T054 - Sample Size "  +  str(size))
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")
    plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("Retail and wherehouse Sale.csv")

sales = df["RETAIL SALES"].dropna()

n = len(sales)
mean = sales.mean()
std = sales.std()

confidence_levels = [0.90, 0.95, 0.99]

print("Mean Retail Sales:", mean)

for confidence in confidence_levels:
    alpha = 1 - confidence
    t_value = stats.t.ppf(1 - alpha/2, n - 1)
    
    margin_error = t_value * (std / np.sqrt(n))
    
    lower = mean - margin_error
    upper = mean + margin_error
    
    print("\nConfidence Level:", int(confidence * 100), "%")
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
```

```python
import pandas as pd
from scipy.stats import ttest_1samp

df = pd.read_csv("Retail and wherehouse Sale.csv")

sales = df["RETAIL SALES"].dropna()

benchmark = 100

t_statistic, p_value = ttest_1samp(sales, benchmark)

print("Business Question:")
print("Is the average Retail Sales significantly different from 100?")

print("\nNull Hypothesis (H0): Mean Retail Sales = 100")
print("Alternative Hypothesis (H1): Mean Retail Sales != 100")
print("Significance Level:", 0.05)

print("\nTest Statistic:", t_statistic)
print("p-value:", p_value)

if p_value < 0.05:
    print("Decision: Reject H0")
    print("Conclusion: Average Retail Sales is significantly different from 100.")
else:
    print("Decision: Do not reject H0")
    print("Conclusion: There is not enough evidence to say that Average Retail Sales is different from 100.")
```

```python
import math
from scipy.stats import norm

visitors_A = 1000
conversions_A = 120

visitors_B = 1000
conversions_B = 150

rate_A = conversions_A / visitors_A
rate_B = conversions_B / visitors_B

pooled_rate = (conversions_A + conversions_B) / (visitors_A + visitors_B)

standard_error = math.sqrt(
    pooled_rate * (1 - pooled_rate) *
    (1 / visitors_A + 1 / visitors_B)
)

z = (rate_A - rate_B) / standard_error

p_value = 2 * (1 - norm.cdf(abs(z)))

print("Version A Conversion Rate:", rate_A * 100, "%")
print("Version B Conversion Rate:", rate_B * 100, "%")

print("\nZ-statistic:", z)
print("p-value:", p_value)
print("Significance Level:", 0.05)

if p_value < 0.05:
    print("Decision: Reject H0")
    print("There is a significant difference between the two conversion rates.")
else:
    print("Decision: Do not reject H0")
    print("There is no significant difference between the two conversion rates.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA

# Load dataset
df = pd.read_csv("Retail and wherehouse Sale.csv")

# Create DATE column
df["DATE"] = pd.to_datetime(
    df["YEAR"].astype(str) + "-" +
    df["MONTH"].astype(str) + "-01"
)

# Calculate monthly retail sales
monthly = df.groupby("DATE")["RETAIL SALES"].sum().sort_index()

print("Monthly Retail Sales:")
print(monthly)

print("\nNumber of Months:", len(monthly))


# -----------------------------
# Time Series Graph
# -----------------------------

plt.figure(figsize=(10,5))
plt.plot(monthly.index, monthly.values, marker="o")
plt.title("Monthly Retail Sales T054")
plt.xlabel("Month")
plt.ylabel("Retail Sales")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()


# -----------------------------
# Prepare data for forecasting
# -----------------------------

sales = monthly.values

# Use simple numeric index
sales_series = pd.Series(sales)


# -----------------------------
# Exponential Smoothing
# -----------------------------

ets_model = ExponentialSmoothing(
    sales_series,
    trend="add",
    seasonal=None
).fit()

ets_forecast = ets_model.forecast(3)

print("\nExponential Smoothing Forecast:")
print(ets_forecast)


# -----------------------------
# ARIMA
# -----------------------------

arima_model = ARIMA(
    sales_series,
    order=(1,1,0)
).fit()

arima_forecast = arima_model.forecast(steps=3)

print("\nARIMA Forecast:")
print(arima_forecast)


# -----------------------------
# Future Month Dates
# -----------------------------

last_date = monthly.index[-1]

future_dates = pd.date_range(
    start=last_date + pd.DateOffset(months=1),
    periods=3,
    freq="MS"
)


# -----------------------------
# Forecast Comparison
# -----------------------------

comparison = pd.DataFrame({
    "Month": future_dates,
    "Exponential Smoothing": ets_forecast.values,
    "ARIMA": arima_forecast.values
})

print("\nForecast Comparison:")
print(comparison)


# -----------------------------
# Forecast Graph
# -----------------------------

plt.figure(figsize=(10,5))

plt.plot(
    monthly.index,
    monthly.values,
    marker="o",
    label="Actual Sales"
)

plt.plot(
    future_dates,
    ets_forecast.values,
    marker="o",
    linestyle="--",
    label="Exponential Smoothing"
)

plt.plot(
    future_dates,
    arima_forecast.values,
    marker="o",
    linestyle="--",
    label="ARIMA"
)

plt.title("Retail Sales Forecast T054")
plt.xlabel("Month")
plt.ylabel("Retail Sales")
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Retail and wherehouse Sale.csv")

# Create date
df["DATE"] = pd.to_datetime(
    df["YEAR"].astype(str) + "-" +
    df["MONTH"].astype(str) + "-01"
)

# Monthly retail sales
monthly = df.groupby("DATE")["RETAIL SALES"].sum().sort_index()

print("Monthly Retail Sales:")
print(monthly)

print("\nNumber of Monthly Records:", len(monthly))


# --------------------------------
# Check number of observations
# --------------------------------

if len(monthly) < 3:
    print("\nNot enough monthly observations for forecast accuracy.")
else:

    # Keep last month as test data
    train = monthly.iloc[:-1]
    test = monthly.iloc[-1:]

    print("\nTraining Data:")
    print(train)

    print("\nActual Test Data:")
    print(test)


    # --------------------------------
    # Simple Forecast
    # --------------------------------

    # Forecast = average of training data
    prediction = train.mean()

    predicted = np.array([prediction])


    # --------------------------------
    # Calculate Errors
    # --------------------------------

    actual = test.values

    mae = np.mean(np.abs(actual - predicted))

    mse = np.mean((actual - predicted) ** 2)

    rmse = np.sqrt(mse)


    # --------------------------------
    # Results
    # --------------------------------

    results = pd.DataFrame({
        "Model": ["Mean Forecast"],
        "MAE": [mae],
        "MSE": [mse],
        "RMSE": [rmse]
    })

    print("\nForecast Accuracy Results:")
    print(results.round(2))


    # --------------------------------
    # Actual vs Predicted
    # --------------------------------

    comparison = pd.DataFrame({
        "Actual Sales": actual,
        "Predicted Sales": predicted
    })

    print("\nActual vs Predicted:")
    print(comparison.round(2))


    # --------------------------------
    # Graph
    # --------------------------------

    plt.figure(figsize=(8,5))

    plt.plot(
        monthly.index,
        monthly.values,
        marker="o",
        label="Actual Sales"
    )

    plt.scatter(
        test.index,
        predicted,
        label="Predicted Sales"
    )

    plt.title("Actual vs Predicted Retail Sales T054")
    plt.xlabel("Month")
    plt.ylabel("Retail Sales")
    plt.legend()
    plt.grid()
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
from scipy.optimize import linprog

# Objective function
# Negative values because linprog minimizes
profit = [-40, -30]

# Constraints
A = [
    [2, 1],
    [1, 2]
]

b = [100, 80]

# Non-negative variables
bounds = [
    (0, None),
    (0, None)
]

# Solve
result = linprog(
    profit,
    A_ub=A,
    b_ub=b,
    bounds=bounds,
    method="highs"
)

# Results
x = result.x[0]
y = result.x[1]

max_profit = -result.fun

machine_used = 2*x + y
labour_used = x + 2*y

print("Product A:", x)
print("Product B:", y)
print("Maximum Profit:", max_profit)

print("Machine Hours Used:", machine_used)
print("Labour Hours Used:", labour_used)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Random demand for 30 days
np.random.seed(10)

demand = np.random.randint(15, 51, 30)

inventory_levels = [30, 40, 50]

results = []

for inventory in inventory_levels:

    total_sales = 0
    total_shortage = 0
    total_leftover = 0

    for daily_demand in demand:

        sales = min(inventory, daily_demand)

        shortage = max(daily_demand - inventory, 0)

        leftover = max(inventory - daily_demand, 0)

        total_sales += sales
        total_shortage += shortage
        total_leftover += leftover

    service_level = total_sales / (total_sales + total_shortage) * 100

    results.append([
        inventory,
        total_sales,
        total_shortage,
        total_leftover,
        service_level
    ])


# Create results table
results_df = pd.DataFrame(
    results,
    columns=[
        "Inventory Level",
        "Total Sales",
        "Total Shortage",
        "Total Leftover",
        "Service Level (%)"
    ]
)

print("Simulation Results:")
print(results_df.round(2))


# Comparison Chart
results_df.plot(
    x="Inventory Level",
    y=["Total Sales", "Total Shortage", "Total Leftover"],
    kind="bar",
    figsize=(9,5)
)

plt.title("Inventory Scenario Comparison T054")
plt.xlabel("Inventory Level")
plt.ylabel("Units")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from scipy.stats import f_oneway

# Load dataset
df = pd.read_csv("Retail and wherehouse Sale.csv")

# Remove missing values
data = df[["ITEM TYPE", "RETAIL SALES"]].dropna()

# Display item types
print("Item Types:")
print(data["ITEM TYPE"].unique())

# Create groups
groups = []

for item_type in data["ITEM TYPE"].unique():
    group = data[data["ITEM TYPE"] == item_type]["RETAIL SALES"]
    groups.append(group)

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(*groups)

print("\nOne-Way ANOVA Results")
print("F-statistic:", f_statistic)
print("p-value:", p_value)
print("Significance Level: 0.05")

# Decision
if p_value < 0.05:
    print("\nDecision: Reject H0")
    print("Conclusion: At least one Item Type has a different mean Retail Sales.")
else:
    print("\nDecision: Do not reject H0")
    print("Conclusion: There is not enough evidence to say that the mean Retail Sales differ among Item Types.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.formula.api import ols

df = pd.read_csv("Retail and wherehouse Sale.csv")

data = df[["ITEM TYPE", "MONTH", "RETAIL SALES"]].dropna()

print("Factors:")
print("1. ITEM TYPE")
print("2. MONTH")

print("\nResponse Variable:")
print("RETAIL SALES")

print("\nAverage Retail Sales:")
print(data.groupby(["ITEM TYPE", "MONTH"])["RETAIL SALES"].mean())

plt.figure(figsize=(10,5))
sns.barplot(
    data=data,
    x="ITEM TYPE",
    y="RETAIL SALES",
    hue="MONTH"
)

plt.title("Retail Sales by Item Type and Month T054")
plt.xlabel("Item Type")
plt.ylabel("Retail Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

model = ols(
    "Q('RETAIL SALES') ~ C(Q('ITEM TYPE')) * C(MONTH)",
    data=data
).fit()

anova = sm.stats.anova_lm(model, typ=2)

print("\nFactorial ANOVA Results:")
print(anova)
```
